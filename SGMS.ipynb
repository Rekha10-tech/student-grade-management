```python

import csv
import os
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
from google.colab import files

class Student:
    SUBJECTS = (
        "Python",
        "DBMS",
        "Mathematics",
        "NLP",
        "Computer Networks"
    )

    def __init__(self, student_id, name, age, marks=None):
        self.student_id = student_id
        self.name = name
        self.age = age
        self.marks = marks if marks else {}

    def calculate_average(self):
        if not self.marks:
            return 0

        return sum(self.marks.values()) / len(self.marks)

    def calculate_grade(self):
        average = self.calculate_average()

        if average >= 90:
            return "A+"
        elif average >= 80:
            return "A"
        elif average >= 70:
            return "B"
        elif average >= 60:
            return "C"
        elif average >= 50:
            return "D"
        else:
            return "F"

    def get_status(self):
        if not self.marks:
            return "Marks Not Added"

        if all(mark >= 40 for mark in self.marks.values()):
            return "PASS"

        return "FAIL"

    def get_report(self):
        return {
            "Student ID": self.student_id,
            "Name": self.name,
            "Age": self.age,
            "Python": self.marks.get("Python", ""),
            "DBMS": self.marks.get("DBMS", ""),
            "Mathematics": self.marks.get("Mathematics", ""),
            "NLP": self.marks.get("NLP", ""),
            "Computer Networks": self.marks.get("Computer Networks", ""),
            "Average": round(self.calculate_average(), 2),
            "Grade": self.calculate_grade(),
            "Status": self.get_status()
        }

class StudentManagementSystem:

    FILE_NAME = "students.csv"

    def __init__(self):
        self.students = []
        self.student_ids = set()
        self.load_from_csv()

    def add_student(self, student):
        if student.student_id in self.student_ids:
            return False, "Student ID already exists."

        self.students.append(student)
        self.student_ids.add(student.student_id)

        return True, "Student added successfully."


    def search_student(self, student_id):
        for student in self.students:
            if student.student_id == student_id:
                return student

        return None



    def delete_student(self, student_id):
        student = self.search_student(student_id)

        if student:
            self.students.remove(student)
            self.student_ids.remove(student_id)

            return True, "Student deleted successfully."

        return False, "Student not found."


    def save_to_csv(self):

        fieldnames = [
            "Student ID",
            "Name",
            "Age",
            "Python",
            "DBMS",
            "Mathematics",
            "NLP",
            "Computer Networks"
        ]

        try:
            with open(self.FILE_NAME, "w",
                      newline="", encoding="utf-8") as file:

                writer = csv.DictWriter(
                    file,
                    fieldnames=fieldnames
                )

                writer.writeheader()

                for student in self.students:

                    row = {
                        "Student ID": student.student_id,
                        "Name": student.name,
                        "Age": student.age
                    }

                    for subject in Student.SUBJECTS:
                        row[subject] = student.marks.get(
                            subject, ""
                        )

                    writer.writerow(row)

            return True, "Data saved successfully to students.csv."

        except Exception as e:
            return False, f"Error saving file: {e}"

    def load_from_csv(self):

        if not os.path.exists(self.FILE_NAME):
            return

        try:

            with open(self.FILE_NAME, "r",
                      newline="", encoding="utf-8") as file:

                reader = csv.DictReader(file)

                for row in reader:

                    marks = {}

                    for subject in Student.SUBJECTS:

                        value = row.get(subject, "")

                        if value != "":
                            try:
                                marks[subject] = float(value)
                            except ValueError:
                                pass

                    try:

                        student = Student(
                            row["Student ID"],
                            row["Name"],
                            int(row["Age"]),
                            marks
                        )

                        self.students.append(student)
                        self.student_ids.add(
                            student.student_id
                        )

                    except (ValueError, KeyError):
                        pass

        except Exception as e:
            print("Error loading CSV:", e)


system = StudentManagementSystem()


title = widgets.HTML(
    """
    <h1 style="text-align:center;
               color:#2c3e50;">
        🎓 Student Grade Management System
    </h1>
    <p style="text-align:center;
              font-size:16px;">
        Python OOP Mini Project
    </p>
    """
)


student_id = widgets.Text(
    description="Student ID:",
    placeholder="Example: S003",
    layout=widgets.Layout(width="450px")
)

name = widgets.Text(
    description="Name:",
    placeholder="Enter student name",
    layout=widgets.Layout(width="450px")
)

age = widgets.IntText(
    description="Age:",
    value=20,
    min=1,
    max=100,
    layout=widgets.Layout(width="450px")
)

python_mark = widgets.FloatText(
    description="Python:",
    value=0,
    min=0,
    max=100,
    layout=widgets.Layout(width="450px")
)

dbms_mark = widgets.FloatText(
    description="DBMS:",
    value=0,
    min=0,
    max=100,
    layout=widgets.Layout(width="450px")
)

math_mark = widgets.FloatText(
    description="Mathematics:",
    value=0,
    min=0,
    max=100,
    layout=widgets.Layout(width="450px")
)

nlp_mark = widgets.FloatText(
    description="NLP:",
    value=0,
    min=0,
    max=100,
    layout=widgets.Layout(width="450px")
)

cn_mark = widgets.FloatText(
    description="Computer Networks:",
    value=0,
    min=0,
    max=100,
    layout=widgets.Layout(width="450px")
)


add_button = widgets.Button(
    description="➕ Add Student",
    button_style="success",
    layout=widgets.Layout(width="150px")
)

update_button = widgets.Button(
    description="✏️ Update Marks",
    button_style="warning",
    layout=widgets.Layout(width="150px")
)

search_button = widgets.Button(
    description="🔍 Search",
    button_style="info",
    layout=widgets.Layout(width="150px")
)

delete_button = widgets.Button(
    description="🗑️ Delete",
    button_style="danger",
    layout=widgets.Layout(width="150px")
)

view_button = widgets.Button(
    description="👥 View All",
    layout=widgets.Layout(width="150px")
)

summary_button = widgets.Button(
    description="📊 Class Summary",
    layout=widgets.Layout(width="150px")
)

save_button = widgets.Button(
    description="💾 Save CSV",
    button_style="primary",
    layout=widgets.Layout(width="150px")
)

download_button = widgets.Button(
    description="⬇️ Download CSV",
    button_style="primary",
    layout=widgets.Layout(width="150px")
)

clear_button = widgets.Button(
    description="🔄 Clear",
    layout=widgets.Layout(width="150px")
)


output = widgets.Output()

def get_marks():

    marks = {
        "Python": python_mark.value,
        "DBMS": dbms_mark.value,
        "Mathematics": math_mark.value,
        "NLP": nlp_mark.value,
        "Computer Networks": cn_mark.value
    }

    for subject, mark in marks.items():

        if mark < 0 or mark > 100:
            raise ValueError(
                f"{subject} mark must be between 0 and 100."
            )

    return marks


def clear_fields():

    student_id.value = ""
    name.value = ""
    age.value = 20

    python_mark.value = 0
    dbms_mark.value = 0
    math_mark.value = 0
    nlp_mark.value = 0
    cn_mark.value = 0


def display_student(student):

    report = student.get_report()

    html = f"""
    <div style="
        border:2px solid #3498db;
        border-radius:10px;
        padding:15px;
        margin:10px;
        background:#f8f9fa;
    ">

    <h3>👤 {report['Name']}</h3>

    <b>Student ID:</b> {report['Student ID']}<br>
    <b>Age:</b> {report['Age']}<br><br>

    <b>Python:</b> {report['Python']}<br>
    <b>DBMS:</b> {report['DBMS']}<br>
    <b>Mathematics:</b> {report['Mathematics']}<br>
    <b>NLP:</b> {report['NLP']}<br>
    <b>Computer Networks:</b> {report['Computer Networks']}<br>

    <hr>

    <b>Average:</b> {report['Average']}<br>
    <b>Grade:</b> {report['Grade']}<br>
    <b>Status:</b> {report['Status']}

    </div>
    """

    display(HTML(html))

def add_student_action(button):

    with output:

        clear_output()

        try:

            if not student_id.value.strip():
                print("❌ Student ID cannot be empty.")
                return

            if not name.value.strip():
                print("❌ Student name cannot be empty.")
                return

            if age.value <= 0:
                print("❌ Invalid age.")
                return

            marks = get_marks()

            new_student = Student(
                student_id.value.strip(),
                name.value.strip(),
                age.value,
                marks
            )

            success, message = system.add_student(
                new_student
            )

            if success:
                print("✅", message)
                display_student(new_student)
            else:
                print("❌", message)

        except Exception as e:
            print("❌ Error:", e)


def update_marks_action(button):

    with output:

        clear_output()

        sid = student_id.value.strip()

        if not sid:
            print("❌ Enter a Student ID.")
            return

        student = system.search_student(sid)

        if not student:
            print("❌ Student not found.")
            return

        try:

            student.marks = get_marks()

            print("✅ Marks updated successfully.")
            display_student(student)

        except Exception as e:
            print("❌ Error:", e)


def search_action(button):

    with output:

        clear_output()

        sid = student_id.value.strip()

        if not sid:
            print("❌ Enter a Student ID.")
            return

        student = system.search_student(sid)

        if student:
            display_student(student)
        else:
            print("❌ Student not found.")



def delete_action(button):

    with output:

        clear_output()

        sid = student_id.value.strip()

        if not sid:
            print("❌ Enter a Student ID.")
            return

        success, message = system.delete_student(sid)

        if success:
            print("✅", message)
        else:
            print("❌", message)


def view_all_action(button):

    with output:

        clear_output()

        if not system.students:
            print("No student records available.")
            return

        print(
            f"📚 Total Students: {len(system.students)}"
        )

        for student in system.students:
            display_student(student)



def summary_action(button):

    with output:

        clear_output()

        if not system.students:
            print("No student records available.")
            return

        grade_count = {}

        total_average = 0

        for student in system.students:

            grade = student.calculate_grade()

            grade_count[grade] = (
                grade_count.get(grade, 0) + 1
            )

            total_average += student.calculate_average()

        class_average = (
            total_average / len(system.students)
        )

        print("📊 CLASS SUMMARY")
        print("=" * 40)

        print(
            f"Total Students : {len(system.students)}"
        )

        print(
            f"Class Average  : {class_average:.2f}"
        )

        print("\nGrade Distribution:")

        for grade, count in sorted(
            grade_count.items()
        ):

            print(
                f"{grade} : {count} student(s)"
            )



def save_action(button):

    with output:

        clear_output()

        success, message = system.save_to_csv()

        if success:
            print("✅", message)
        else:
            print("❌", message)



def download_action(button):

    with output:

        clear_output()

        success, message = system.save_to_csv()

        if success:

            print(
                "✅ CSV saved. Downloading..."
            )

            files.download(
                system.FILE_NAME
            )

        else:
            print("❌", message)



def clear_action(button):

    clear_fields()

    with output:
        clear_output()

        print("🔄 Input fields cleared.")



add_button.on_click(add_student_action)
update_button.on_click(update_marks_action)
search_button.on_click(search_action)
delete_button.on_click(delete_action)
view_button.on_click(view_all_action)
summary_button.on_click(summary_action)
save_button.on_click(save_action)
download_button.on_click(download_action)
clear_button.on_click(clear_action)



input_section = widgets.VBox([
    student_id,
    name,
    age,
    widgets.HTML("<hr><b>Enter Marks (0-100)</b>"),
    python_mark,
    dbms_mark,
    math_mark,
    nlp_mark,
    cn_mark
])

button_row1 = widgets.HBox([
    add_button,
    update_button,
    search_button,
    delete_button
])

button_row2 = widgets.HBox([
    view_button,
    summary_button,
    save_button,
    download_button,
    clear_button
])

display(
    widgets.VBox([
        title,
        widgets.HTML("<h3>Student Details</h3>"),
        input_section,
        widgets.HTML("<br>"),
        button_row1,
        button_row2,
        widgets.HTML("<hr>"),
        output
    ])
)
```
